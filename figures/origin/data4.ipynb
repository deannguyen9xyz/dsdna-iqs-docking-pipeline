```python
import pandas as pd
import os

df = pd.read_csv(os.path.join("..", "generated_sequences", "stage3_for_docking.csv"))
df['gc_percent'] = df['gc_content'] * 100

print(f"Mean GC  : {df['gc_percent'].mean():.1f}%")
print(f"Min GC   : {df['gc_percent'].min():.1f}%")
print(f"Max GC   : {df['gc_percent'].max():.1f}%")
print(f"Std dev  : {df['gc_percent'].std():.1f}%")

# Count sequences in key ranges
below_65 = (df['gc_percent'] < 65).sum()
above_70  = (df['gc_percent'] >= 70).sum()
print(f"Below 65%: {below_65} sequences")
print(f"Above 70%: {above_70} sequences")
```

```python
import pandas as pd
import os

df = pd.read_csv(os.path.join("..", "generated_sequences", "stage3_for_docking.csv"))

gg_counts = df['gg_steps'].value_counts().sort_index()
print("GG steps distribution for Origin:")
print()
print("GG_steps\tCount")
for gg, count in gg_counts.items():
    print(f"{gg}\t{count}")

print()
print(f"Mean GG steps : {df['gg_steps'].mean():.2f}")
print(f"Min GG steps  : {df['gg_steps'].min()}")
print(f"Max GG steps  : {df['gg_steps'].max()}")
```

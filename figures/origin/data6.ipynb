```python
import pandas as pd
import os

# Load final 100
df_final = pd.read_csv(os.path.join("..", "generated_sequences", "stage3_for_docking.csv"))
df_final['gc_percent'] = df_final['gc_content'] * 100

# Load Stage 1 (all 10,000)
df_stage1 = pd.read_csv(os.path.join("..", "generated_sequences", "stage1_candidates.csv"))
df_stage1['gc_percent'] = df_stage1['gc_content'] * 100

print("=== ALL 10,000 (Stage 1) ===")
print(f"GC content (%)    : {df_stage1['gc_percent'].mean():.1f}")
print(f"GG steps          : {df_stage1['gg_steps'].mean():.2f}")
print(f"Purine fraction   : {df_stage1['purine_fraction'].mean():.3f}")
print(f"Length (bases)    : {df_stage1['length'].mean():.1f}")
print()
print("=== FINAL 100 ===")
print(f"GC content (%)    : {df_final['gc_percent'].mean():.1f}")
print(f"GG steps          : {df_final['gg_steps'].mean():.2f}")
print(f"Purine fraction   : {df_final['purine_fraction'].mean():.3f}")
print(f"Length (bases)    : {df_final['length'].mean():.1f}")
```

```python
properties = {
    'GC content (%)':    (df_stage1['gc_percent'].mean(),   df_final['gc_percent'].mean()),
    'GG steps':          (df_stage1['gg_steps'].mean(),     df_final['gg_steps'].mean()),
    'Purine fraction':   (df_stage1['purine_fraction'].mean(), df_final['purine_fraction'].mean()),
    'Length (bases)':    (df_stage1['length'].mean(),       df_final['length'].mean()),
}

print("Property           Stage1    Final100   Change(%)")
print("-" * 55)
for prop, (s1, f1) in properties.items():
    change = ((f1 - s1) / s1) * 100
    print(f"{prop:<22} {s1:>8.2f}  {f1:>8.2f}   {change:>+.1f}%")
```
